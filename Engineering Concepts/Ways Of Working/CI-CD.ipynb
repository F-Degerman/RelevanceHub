**CI/CD** stands for **Continuous Integration** and **Continuous Delivery**, or sometimes **Continuous Deployment**.

Simply explained:

- **Continuous Integration (CI)**
  means that code changes are added to the project frequently and automatically tested to make sure nothing has broken.
  Example:
  when you push code, tests, linting, or a build run automatically.

- **Continuous Delivery (CD)**
  means that the code is automatically prepared for release or deployment.
  In other words, it can be ready to ship at any time.

- **Continuous Deployment**
  goes one step further:
  if everything passes the tests, the code can be automatically released all the way to production without any manual button press.

In short:
- **CI** = automatically verify that new code works together with existing code
- **CD** = make it easy and fast to deliver or deploy new code


A common example of a CI/CD workflow:
1. You push code to GitHub
2. GitHub Actions starts automatically
3. Tests run
4. Code quality is checked
5. If everything succeeds, the app can be built and possibly deployed

So CI/CD is not a development framework like Scrum, but rather a way of working and an automated process for building, testing, and delivering code.
