# Linux & Git Cheat Sheet

## Navigation

| Command | Description |
|-----------|-------------|
| `pwd` | Show the current directory |
| `ls` | List the contents of the current directory |
| `cd directory-name` | Change directory |
| `cd ..` | Go up one level |
| `cd ~` | Go to the home directory |

## Create directories and files

| Command | Description |
|-----------|-------------|
| `mkdir directory-name` | Create a directory |
| `mkdir -p project/data-engineering` | Create nested directories as needed |
| `touch file-name.txt` | Create an empty file |

## VS Code

| Command | Description |
|-----------|-------------|
| `code .` | Open the current directory in VS Code |

## System

| Command | Description |
|-----------|-------------|
| `sudo apt update` | Update the package list |
| `sudo apt upgrade -y` | Upgrade installed packages |
| `sudo apt install git -y` | Install Git |

## Check installations

| Command | Description |
|-----------|-------------|
| `git --version` | Show the Git version |
| `python3 --version` | Show the Python version |
| `which git` | Show where Git is installed |
| `which python3` | Show where Python is installed |

## Git

| Command | Description |
|-----------|-------------|
| `git clone <repo-url>` | Clone a repository |
| `git status` | Show the status |
| `git add .` | Stage all changes |
| `git commit -m "Message"` | Create a commit |
| `git push` | Push to GitHub |

## .gitignore

| Entry in `.gitignore` | Description |
|------------------------|-------------|
| `private/` | Ignore the private directory |
| `.vscode/` | Ignore VS Code settings |
| `.env` | Ignore environment variables and secrets |
