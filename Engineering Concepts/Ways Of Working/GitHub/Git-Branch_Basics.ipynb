# Git Branch Workflow in VS Code, Git Graph & GitHub


## 1. The basic idea

A **branch** is an independent line of development.

Normally, **main** contains the stable version of the project. Instead of making changes directly on **main**, you create a separate branch.

For example:

```text
main
  │
  ●  Initial commit
  │
  ●  Application source files
  │
  ●  Project setup documentation
  │
  └───────────────► feature/documentation-update
```

You can then work safely on **feature/documentation-update** without changing **main**.

---


# Part 1 — Create a branch


## 2. Check your current branch

In VS Code, look at the **bottom-left status bar**.

You should see something like:

```text
⑂ main
```

This means that your current branch is **main**.


## 3. Create a new branch

Click the branch name (**main**) in the status bar.

Choose:

**Create new branch...**

Enter a name, for example:

```text
feature/documentation-update
```

VS Code automatically switches to the new branch.

The status bar should now show:

```text
⑂ feature/documentation-update
```

At this point, both branches point to exactly the same commit:

```text
                  main
                  feature/documentation-update
                     ↓
●────●────●─────────●
```

There is no visible split yet because nothing has been committed to **feature/documentation-update**.

---


# Part 2 — Make a change


## 4. Edit a tracked file

Change a file in the project while you are on **feature/documentation-update**.

For example, add some Markdown:

```markdown
# Feature documentation update

Testing Git branches.
```

Save the file with **Ctrl + S**.


### Important: **.gitignore**

If the file or its directory is ignored by **.gitignore**, Git will normally not track a new ignored file.

For example:

```text
project/
├── README.md              ← Git tracks this
├── notebook.ipynb         ← Git tracks this
├── .gitignore
│
└── ignored_folder/
    └── notes.md           ← ignored
```

For a branch change to be committed, Git needs to be tracking the relevant file/change.

---


# Part 3 — Review the change


## 5. Open Source Control

Click the **Source Control** icon in VS Code.

Your modified file should appear under:

**Changes**

You can click the file to inspect the difference.

VS Code may show something similar to:

```text
OLD VERSION                 NEW VERSION

                            + Testing Git branches
```

Green generally represents something that has been **added**.

Reviewing your changes before committing is a good habit.

---


# Part 4 — Stage the change


## 6. Stage the file

In **Source Control**, hover over the changed file and click:

****+****

The file moves from:

```text
Changes
```

to:

```text
Staged Changes
```

The concept is:

```text
Working directory
      │
      │  Stage (+)
      ▼
Staging area
      │
      │  Commit
      ▼
Git history
```

**Staging** means:

> Include this change in my next commit.

This is useful because you might have modified five files but only want three of them in a particular commit.

---


# Part 5 — Commit


## 7. Write a commit message

In Source Control, enter a short description of the change.

For example:

```text
Add workflow documentation
```

Then click:

**Commit**

The commit is now stored in your **local Git repository**.

It has **not necessarily been sent to GitHub yet**.

Now your history looks roughly like this:

```text
                         feature/documentation-update
                              ↓
                              ●  Add workflow documentation
                             /
●────●────●────────────────●
                         ↑
                        main
```

This is the first point where the branches actually contain different histories.

**main** still points to the old commit.

**feature/documentation-update** points to your new commit.

---


# Part 6 — View it in Git Graph


## 8. Open Git Graph

In VS Code:

**Ctrl + Shift + P**

Search for:

```text
Git Graph: View Git Graph
```

Git Graph visualizes your repository.

At this stage you may see:

```text
●  feature/documentation-update    Add workflow documentation
│
●  main           Project setup documentation
│
●                 files
│
●                 Create project structure
│
●                 Initial commit
```

The labels are pointers.

For example:

```text
feature/documentation-update ──────► ●
                    │
main ─────────────► ●
                    │
                    ●
```

The branches are pointing at different commits.

---


# Part 7 — Publish the branch to GitHub


## 9. Push / Publish Branch

So far, **feature/documentation-update** may exist only on your computer.

In Source Control, choose:

**Publish Branch**

This pushes the branch to GitHub.

Conceptually:

```text
LOCAL COMPUTER                    GITHUB

feature/documentation-update                       feature/documentation-update
     ● ─────────── PUSH ─────────► ●
     │                             │
main ●                       main  ●
```

Afterward, Git Graph may show both a local and remote reference.


### What is **origin**?

**origin** is the conventional name Git gives to the remote repository you cloned from — in this case, your GitHub repository.

You can think of it approximately as:

```text
main
    = my local main

origin/main
    = my local remote-tracking reference for main on GitHub
```

---


# Part 8 — Create a Pull Request


## 10. Open GitHub

Go to the repository on GitHub.

After recently publishing the branch, GitHub will often offer:

**Compare & pull request**

Create a Pull Request with:

```text
base: main  ←  compare: feature/documentation-update
```

This means:

> I want to merge the changes from **feature/documentation-update** into **main**.

Before creating or merging the Pull Request, GitHub lets you inspect:

```text
Conversation
Commits
Checks
Files changed
```

This is especially useful in group projects.

A teammate can review your code before it enters **main**.

---


# Part 9 — Merge the Pull Request


## 11. Check for conflicts

GitHub may report:

**No conflicts with base branch**

This means GitHub can automatically combine the branches.

Then choose:

**Merge pull request**

and:

**Confirm merge**

The important point is that the merge occurs **on GitHub**.

Your local repository does not automatically update just because something changed on GitHub.

After the merge, the history can look like:

```text
               ●  origin/main
              / \
             /   \
            /     ●  feature/documentation-update
           /     /
●─────────●─────●
          ↑
     local main
```

This is very similar to what we saw in Git Graph.

The merge commit on GitHub is now newer than your local **main**.

---


# Part 10 — Fetch


## 12. Fetch information from GitHub

In VS Code:

**Ctrl + Shift + P**

Run:

```text
Git: Fetch
```

**fetch** downloads information about new commits and branches from the remote repository **without switching your branch or directly integrating those commits into your current branch**.

After our fetch, Git Graph showed roughly:

```text
●  origin/main     Merge pull request #1...
│\
│ ● feature/documentation-update    Add workflow documentation
│/
●  main            Project setup documentation
│
●                  files
│
●                  Create project structure
│
●                  Initial commit
```

This visualization is very useful.

It tells us:

```text
origin/main  ─────► NEW
                     ●
                    / \
                   /   ● ◄──── feature/documentation-update
                  /
main ───────────► ●
```

Git now knows that GitHub's **main** has moved forward.

But our local **main** has not caught up yet.

---


# Part 11 — Switch back to main


## 13. Checkout **main**

Click the branch name in the VS Code status bar.

Switch from:

```text
feature/documentation-update
```

to:

```text
main
```

The status bar should now show:

```text
⑂ main
```

---


# Part 12 — Pull


## 14. Pull the new version of **main**

Run:

**Ctrl + Shift + P → **Git: Pull****

Pull brings the remote changes into your local branch.

In our example:


### Before Pull

```text
origin/main ─────► ●
                  |\
                  | ● feature/documentation-update
                  |/
main ────────────►●
```


### After Pull

```text
main ────────────┐
origin/main ─────┴──► ●
                     |\
                     | ● feature/documentation-update
                     |/
                     ●
```

Now your local **main** and GitHub's **main** are synchronized.

That is exactly what we saw at the end in Git Graph.

---


# Fetch vs Pull vs Push vs Sync Changes

These four concepts are worth learning separately.

| Command          | Direction                      | Meaning                            |
| ---------------- | ------------------------------ | ---------------------------------- |
| **Fetch**        | GitHub → local Git information | Find out what has changed remotely |
| **Pull**         | GitHub → local branch          | Get and integrate remote changes   |
| **Push**         | Local → GitHub                 | Upload your commits                |
| **Sync Changes** | Both directions                | Pull + Push to synchronize         |

A useful mental model is:

```text
                    FETCH
             ┌──────────────────
             │
             ▼
        ┌─────────┐
        │  LOCAL  │
        │   GIT   │
        └─────────┘
             ▲ │
        PULL │ │ PUSH
             │ ▼
        ┌─────────┐
        │ GITHUB  │
        └─────────┘
```

More precisely, **fetch** updates your local knowledge of remote references such as **origin/main**; **pull** then integrates remote work into your current local branch.

---


# The complete workflow

The complete workflow we practiced was:

```text
1. Start on main
       │
       ▼
2. Create feature/documentation-update
       │
       ▼
3. Edit a file
       │
       ▼
4. Review Changes
       │
       ▼
5. Stage (+)
       │
       ▼
6. Commit
       │
       ▼
7. Publish / Push
       │
       ▼
8. Create Pull Request
       │
       ▼
9. Review changes
       │
       ▼
10. Merge into main on GitHub
       │
       ▼
11. Fetch
       │
       ▼
12. Switch to local main
       │
       ▼
13. Pull
       │
       ▼
14. Local main = GitHub main
```


## How this works in a group

Suppose three people are working on a data engineering project:

```text
                         ┌── feature/api
                         │
main ────────────────────┼── feature/airflow
                         │
                         └── feature/database
```

Each person works independently.

Later:

```text
feature/api ──────── PR ───┐
                            │
feature/airflow ──── PR ────┼──► main
                            │
feature/database ─── PR ────┘
```

This prevents everyone from constantly editing **main** directly.

For a data engineering project, branches could for example be named:

```text
feature/api-integration
feature/data-processing
feature/workflow-automation
feature/database
fix/docker-compose
docs/readme
```

That workflow also connects well to the kind of Airflow/data-pipeline work in your course, where different pipeline components can naturally be developed as separate changes before integration.


### The key mental model

You do **not** really "put files into a branch."

A branch is a **pointer to a sequence of commits**.

The most useful way to think about the workflow is:

```text
EDIT
  ↓
STAGE
"What goes into the next commit?"
  ↓
COMMIT
"Save this snapshot locally."
  ↓
PUSH
"Send my commits to GitHub."
  ↓
PULL REQUEST
"Please review/integrate my branch."
  ↓
MERGE
"Combine it with main."
  ↓
PULL
"Update my local main."
```

Once that distinction is clear, most of Git becomes much easier to reason about.
